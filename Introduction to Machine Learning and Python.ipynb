## 2️⃣ Machine Learning..




> Machine Learning is a subset of AI where computers learn patterns from data instead of being explicitly programmed. Instead of writing every rule, we provide data and the machine learns the pattern.



**Example: Spam Email Detection**

1. **Input**: Thousands of emails
2. **Process**: Machine learns patterns
3. **Predicts**: Spam or Not Spam

## 3️⃣ AI vs ML vs Data Science

| AI | Machine Learning | Data Science |
| :--- | :--- | :--- |
| Makes machines intelligent | Learns from data | Extracts insights from data |
| Broad field | Subset of AI | Uses AI + Statistics |
| Example: Robot | Example: Recommendation | Example: Sales Analysis |

## 4️⃣ Types of Machine Learning

### 1. Supervised Learning
Data contains Input + Output (Labels).
* **Example**: Predict House Price
* **Algorithms**: Linear Regression, Logistic Regression, Decision Tree, Random Forest

### 2. Unsupervised Learning
Only Input, no output labels. Goal is to find hidden patterns.
* **Example**: Customer Segmentation
* **Algorithms**: K-Means, PCA, Hierarchical Clustering

### 3. Reinforcement Learning
Machine learns by Reward and Punishment.
* **Example**: Chess AI, Robots, Self Driving Cars

## 8️⃣ Matplotlib

Standard library for creating static, animated, and interactive visualizations.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.show()
```
